## Task 1 — Decimal, Binary & Hexadecimal
Take an integer and display its decimal, binary, and hexadecimal representations.

```python
number = int(input("Enter an integer: "))

print("Decimal    :", number)
print("Binary     :", bin(number))
print("Hexadecimal:", hex(number))
```

## Task 2 — ASCII Value Conversion
Take a character, display its ASCII value, then convert the ASCII value back to a character.

```python
character = input("Enter a single character: ")

ascii_value = ord(character)
print(f"Character '{character}' → ASCII value: {ascii_value}")

back_to_char = chr(ascii_value)
print(f"ASCII value {ascii_value} → Character: '{back_to_char}'")
```

## Task 3 — UTF-8 Encoding & Decoding
Take a message, encode it using UTF-8, then decode it back to text.

```python
message = input("Enter a message: ")

encoded = message.encode("utf-8")
print("Original :", message)
print("Encoded  :", encoded)
print("Type     :", type(encoded))

decoded = encoded.decode("utf-8")
print("Decoded  :", decoded)
```

## Task 4 — Password Strength Checker
Check whether a password meets these requirements:
- At least 8 characters
- At least one digit
- At least one uppercase letter

```python
password = input("Enter your password: ")

has_length = len(password) >= 8
has_digit  = any(char.isdigit() for char in password)
has_upper  = any(char.isupper() for char in password)

print("\n--- Password Check ---")
print(f"At least 8 characters : {'✓' if has_length else '✗'}")
print(f"Contains a digit      : {'✓' if has_digit  else '✗'}")
print(f"Contains uppercase    : {'✓' if has_upper  else '✗'}")

if has_length and has_digit and has_upper:
    print("\nPassword meets the basic security requirements.")
else:
    print("\nPassword does NOT meet the security requirements.")
```

## Task 5 — Login Checker (Max 3 Attempts)
Allow a user to log in with a maximum of three attempts.

```python
CORRECT_USERNAME = "admin"
CORRECT_PASSWORD = "Admin123"
MAX_ATTEMPTS     = 3

attempt = 0
logged_in = False

while attempt < MAX_ATTEMPTS:
    attempt += 1
    print(f"\n--- Attempt {attempt} of {MAX_ATTEMPTS} ---")
    username = input("Username: ")
    password = input("Password: ")

    if username == CORRECT_USERNAME and password == CORRECT_PASSWORD:
        print("Access Granted! Welcome,", username)
        logged_in = True
        break
    else:
        remaining = MAX_ATTEMPTS - attempt
        if remaining > 0:
            print(f"Incorrect credentials. {remaining} attempt(s) remaining.")
        else:
            print("Access Denied. Maximum attempts reached. Account locked.")
```

## Challenge — Login Checker + Password Strength Validation
Combine login verification (max 3 attempts) with password strength requirements from Task 4.

The user must provide both valid credentials **and** a password that meets all strength rules.

```python
CORRECT_USERNAME = "admin"
CORRECT_PASSWORD = "Admin123"
MAX_ATTEMPTS     = 3

def check_password_strength(pwd):
    """Returns (is_strong, list_of_issues)"""
    issues = []
    if len(pwd) < 8:
        issues.append("at least 8 characters")
    if not any(c.isdigit() for c in pwd):
        issues.append("at least one digit")
    if not any(c.isupper() for c in pwd):
        issues.append("at least one uppercase letter")
    return len(issues) == 0, issues

attempt   = 0
logged_in = False

while attempt < MAX_ATTEMPTS:
    attempt += 1
    print(f"\n--- Attempt {attempt} of {MAX_ATTEMPTS} ---")
    username = input("Username: ")
    password = input("Password: ")

    # First check password strength
    is_strong, issues = check_password_strength(password)
    if not is_strong:
        print("Weak password. Requirements not met:")
        for issue in issues:
            print(f"  • Missing: {issue}")
        remaining = MAX_ATTEMPTS - attempt
        if remaining > 0:
            print(f"{remaining} attempt(s) remaining.")
        else:
            print("Maximum attempts reached. Access Denied.")
        continue

    # Then verify credentials
    if username == CORRECT_USERNAME and password == CORRECT_PASSWORD:
        print("Password is strong. Credentials verified.")
        print(f"Access Granted! Welcome, {username}!")
        logged_in = True
        break
    else:
        remaining = MAX_ATTEMPTS - attempt
        print("Incorrect credentials.", end=" ")
        if remaining > 0:
            print(f"{remaining} attempt(s) remaining.")
        else:
            print("Maximum attempts reached. Access Denied.")
```
